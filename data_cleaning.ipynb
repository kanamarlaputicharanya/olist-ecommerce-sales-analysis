```python
import unicodedata
```

```python
def remove_accents(text):
    return ''.join(
        c for c in unicodedata.normalize('NFKD', text)
        if not unicodedata.combining(c)
    )
```

```python
geolocation_clean["geolocation_city"] = (
    geolocation_clean["geolocation_city"]
    .str.lower()
    .str.strip()
    .apply(remove_accents)
)
```
